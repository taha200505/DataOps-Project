### 1. Generate the customer dataset

```python
import csv
import random
from datetime import date, timedelta
from pathlib import Path
```

```python
random.seed(42)

OUTPUT_DIR = Path("data")
OUTPUT_FILE = OUTPUT_DIR / "customers.csv"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

cities = {
    "Maharashtra": ["Pune", "Mumbai", "Nagpur"],
    "Karnataka": ["Bangalore", "Mysore", "Mangalore"],
    "Telangana": ["Hyderabad", "Warangal"],
    "Tamil Nadu": ["Chennai", "Coimbatore", "Madurai"],
    "Delhi": ["New Delhi"],
    "Gujarat": ["Ahmedabad", "Surat", "Vadodara"],
}

columns = [
    "customer_id",
    "name",
    "city",
    "state",
    "country",
    "registration_date",
    "is_active",
]
```

```python
def random_date(start_year=2022, end_year=2025):
    start = date(start_year, 1, 1)
    end = date(end_year, 12, 31)
    days = (end - start).days
    return (start + timedelta(days=random.randint(0, days))).isoformat()
```

```python
OUTPUT_ROWS = 10000

with OUTPUT_FILE.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()

    for customer_id in range(OUTPUT_ROWS):
        state = random.choice(list(cities.keys()))
        city = random.choice(cities[state])

        writer.writerow({
            "customer_id": str(customer_id),
            "name": f"Customer_{customer_id}",
            "city": city,
            "state": state,
            "country": "India",
            "registration_date": random_date(),
            "is_active": str(random.choice([True, False])),
        })

print(f"Generated {OUTPUT_ROWS} customers: {OUTPUT_FILE}")
```

### 2. Define the Spark schema — all columns as strings

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType

spark = SparkSession.builder \
    .appName("CustomerDataPipeline") \
    .getOrCreate()

spark
```

```python
# Define Schema
customer_schema = StructType([
    StructField("customer_id", StringType(), True),
    StructField("name", StringType(), True),
    StructField("city", StringType(), True),
    StructField("state", StringType(), True),
    StructField("country", StringType(), True),
    StructField("registration_date", StringType(), True),
    StructField("is_active", StringType(), True),
])
```

```python
print(customer_schema.simpleString())
```

### 4. Read the CSV from HDFS using your schema

```python
raw_path = "hdfs://namenode:9000/data/customers/raw/customers.csv"

df = spark.read \
    .option("header", True) \
    .option("mode", "PERMISSIVE") \
    .schema(customer_schema) \
    .csv(raw_path)


df.show(5, truncate=False)
df.printSchema()
print("Total rows:", df.count())
```

### 5. Transform and clean the data with Spark

```python

from pyspark.sql import functions as F

clean_df =df.withColumn("name", F.trim(F.col("name"))) \
    .withColumn("city", F.trim(F.col("city"))) \
    .withColumn("state", F.trim(F.col("state"))) \
    .withColumn("customer_id",F.col("customer_id").cast("integer")) \
    .withColumn("registration_date",F.to_date(F.col("registration_date"), "yyyy-MM-dd")) \
    .withColumn("is_active",F.lower(F.trim(F.col("is_active"))).cast("boolean") ) \
    .filter(F.col("customer_id").isNotNull())

clean_df.printSchema()
clean_df.show(5, truncate=False)
```

### 6. Save the processed data to HDFS as Parquet

```python
output_path = "hdfs://namenode:9000/data/customers/processed"
clean_df.write.mode("overwrite").parquet(output_path)

print("Processed data saved to HDFS:", output_path)
```

```python
processed_df = spark.read.parquet(output_path)

processed_df.show(10, truncate=False)
print("Processed rows:", processed_df.count())
```
