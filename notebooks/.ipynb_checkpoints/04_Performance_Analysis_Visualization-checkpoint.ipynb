```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### <span style="background-color: yellow;">Chargement des données et calcul des métriques de performance</span>

```python
# Configuration du style graphique professionnel
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context("notebook", font_scale=1.1)

# 1. Chargement des résultats depuis le fichier JSON
chemin_json = "/home/jovyan/work/benchmark_results.json"
with open(chemin_json, "r") as f:
    raw_results = json.load(f)

# 2. Création d'un DataFrame Pandas structuré
df_bench = pd.DataFrame(list(raw_results.items()), columns=["Strategie", "Temps_Execution_s"])

# Extraction du numéro de niveau pour garantir un tri chronologique strict (1 à 10)
df_bench["Niveau"] = df_bench["Strategie"].apply(lambda x: int(x.split(".")[0]))
df_bench = df_bench.sort_values("Niveau").reset_index(drop=True)

# 3. Calcul des indicateurs de performance (KPIs) par rapport à la Baseline (Niveau 1)
temps_baseline = df_bench.loc[0, "Temps_Execution_s"]

# Speedup : Combien de fois la stratégie est plus rapide que la Baseline (ex: 5.0x)
df_bench["Speedup_x"] = round(temps_baseline / df_bench["Temps_Execution_s"], 2)

# Gain en pourcentage (%) : Réduction du temps de calcul par rapport à la Baseline
df_bench["Gain_Pourcentage"] = round(((temps_baseline - df_bench["Temps_Execution_s"]) / temps_baseline) * 100, 1)

# Affichage du tableau récapitulatif
print("📊 TABLEAU RÉCAPITULATIF DU BENCHMARK SPARK :")
display(df_bench[["Niveau", "Strategie", "Temps_Execution_s", "Speedup_x", "Gain_Pourcentage"]])
```

#### <span style="background-color: yellow;">Comparaison des temps d'exécution (Bar Chart Horizontal)</span>

```python
# Création d'une figure haute résolution
fig, ax = plt.subplots(figsize=(12, 7))

# Palette de couleurs dégradée (du rouge pour le plus lent au vert pour le plus rapide)
colors = sns.color_palette("RdYlGn_r", len(df_bench))
# Trier les couleurs selon le temps d'exécution réel
ranks = df_bench["Temps_Execution_s"].rank().astype(int) - 1
bar_colors = [colors[r] for r in ranks]

# Tracé des barres horizontales
bars = ax.barh(df_bench["Strategie"], df_bench["Temps_Execution_s"], color=bar_colors, edgecolor="black", height=0.65)

# Inversion de l'axe Y pour avoir le Niveau 1 en haut et le Niveau 10 en bas
ax.invert_yaxis()

# Ajout des annotations (Temps exact + Speedup) au bout de chaque barre
for bar, speedup in zip(bars, df_bench["Speedup_x"]):
    width = bar.get_width()
    ax.text(
        width + (temps_baseline * 0.015), # Décalage léger à droite de la barre
        bar.get_y() + bar.get_height() / 2,
        f"{width:.2f} s  ({speedup}x)",
        va="center",
        ha="left",
        fontsize=10,
        fontweight="bold",
        color="#222222"
    )

# Habillage du graphique
ax.set_title("Benchmark Apache Spark : Temps d'Exécution par Stratégie d'Optimisation", fontsize=14, fontweight="bold", pad=15)
ax.set_xlabel("Temps d'exécution (en secondes) — Plus bas = Meilleur", fontsize=12, fontweight="bold")
ax.set_ylabel("Stratégies d'Optimisation", fontsize=12, fontweight="bold")
ax.set_xlim(0, temps_baseline * 1.18) # Marge à droite pour le texte

plt.tight_layout()
plt.savefig("/home/jovyan/work/graph1_temps_execution.png", dpi=300)
plt.show()
```

#### <span style="background-color: yellow;">Cellule 3 : Graphique 2 — Facteur d'Accélération (Speedup) et Gain (%)</span>

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6), sharey=True)

# --- Sous-graphique 1 : Facteur d'Accélération (Speedup) ---
sns.barplot(
    data=df_bench,
    x="Speedup_x",
    y="Strategie",
    palette="Blues_d",
    ax=ax1,
    edgecolor="black"
)
ax1.axvline(1.0, color="red", linestyle="--", linewidth=1.2, label="Baseline (1.0x)")
ax1.set_title("Facteur d'Accélération (Speedup vs Baseline)", fontsize=13, fontweight="bold")
ax1.set_xlabel("Multiplicateur de vitesse (X fois plus rapide)", fontsize=11)
ax1.set_ylabel("")
ax1.legend()

for i, v in enumerate(df_bench["Speedup_x"]):
    ax1.text(v + 0.1, i, f"{v}x", va="center", fontweight="bold", fontsize=9.5)

# --- Sous-graphique 2 : Réduction du temps en Pourcentage (%) ---
sns.barplot(
    data=df_bench,
    x="Gain_Pourcentage",
    y="Strategie",
    palette="Greens_d",
    ax=ax2,
    edgecolor="black"
)
ax2.set_title("Réduction du Temps de Calcul (%)", fontsize=13, fontweight="bold")
ax2.set_xlabel("Gain de performance (%)", fontsize=11)
ax2.set_ylabel("")
ax2.set_xlim(0, 110)

for i, v in enumerate(df_bench["Gain_Pourcentage"]):
    ax2.text(v + 1.5, i, f"{v}%", va="center", fontweight="bold", fontsize=9.5)

plt.suptitle("Impact Quantitatif des Optimisations sur le Cluster Spark (2 Workers / 2 Go RAM)", fontsize=15, fontweight="bold", y=1.02)
plt.tight_layout()
plt.savefig("/home/jovyan/work/graph2_speedup_gains.png", dpi=300)
plt.show()
```
